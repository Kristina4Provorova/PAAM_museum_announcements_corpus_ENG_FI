# **Загрузка инструмента «Pipeline API» и моделей OPUS-MT и M2M-100**

Репозиторий инструмента "Pipeline API": https://huggingface.co/docs/transformers.js/pipelines

Репозиторий модели OPUS-MT: https://huggingface.co/Helsinki-NLP/opus-mt-ru-en

Репозиторий модели M2M-100: https://huggingface.co/docs/transformers/model_doc/m2m_100

```python
# ============================================================
# ЗАНЯТИЕ 3: Две модели, source.txt, выгрузка переводов
# ============================================================
!pip install "transformers<5.0.0" sentencepiece sacrebleu bert-score pandas -q

from transformers import pipeline

# ============================================================
# АЛГОРИТМ (совпадает с прошлыми занятиями):
# Загрузка библиотек → Импорт → Создание объекта → Чтение файла → Перевод → Сохранение
# ============================================================

# --- ШАГ 3: Создание объектов (две модели) ---

# OPUS-MT: одна пара (ru→en)
opus = pipeline("translation_ru_to_en", model="Helsinki-NLP/opus-mt-ru-en", device="cuda")

# M2M: мультиязычная
m2m = pipeline("translation", model="facebook/m2m100_418M", src_lang="ru", tgt_lang="en", device="cuda")

# --- ШАГ 4: Чтение файла ---
with open("source_text.txt", "r", encoding="utf-8") as f:# ЗДЕСЬ НЕОБХОДИМО БУДЕТ МЕНЯТЬ ФАЙЛ ДЛЯ ЗАДАНИЯ 1!!!!!!!!!!!!!
    text = f.read()

# --- ШАГ 5: Перевод ---

# OPUS-MT: Greedy
opus_greedy = opus(text)[0]["translation_text"]

# OPUS-MT: Beam Search
opus_beam = opus(text, num_beams=5)[0]["translation_text"]

# OPUS-MT: Семплирование
opus_sample = opus(text, do_sample=True, temperature=0.7)[0]["translation_text"]

# M2M: нужно указать языки
m2m.tokenizer.src_lang = "ru"
# M2M: Greedy
m2m_greedy = m2m(text)[0]["translation_text"]
# M2M: Beam Search
m2m_beam = m2m(text, num_beams=5)[0]["translation_text"]
# M2M: Семплирование
m2m_sample = m2m(text, do_sample=True, temperature=0.7)[0]["translation_text"]

# --- ШАГ 6: Сохранение ---
with open("opus_greedy.txt", "w", encoding="utf-8") as f: f.write(opus_greedy)
with open("opus_beam.txt", "w", encoding="utf-8") as f: f.write(opus_beam)
with open("opus_sample.txt", "w", encoding="utf-8") as f: f.write(opus_sample)
with open("opus_sample_1.2.txt", "w", encoding="utf-8") as f: f.write(opus_sample_1_2)

with open("m2m_greedy.txt", "w", encoding="utf-8") as f: f.write(m2m_greedy)
with open("m2m_beam.txt", "w", encoding="utf-8") as f: f.write(m2m_beam)
with open("m2m_sample.txt", "w", encoding="utf-8") as f: f.write(m2m_sample)
with open("m2m_sample_1.2.txt", "w", encoding="utf-8") as f: f.write(m2m_sample_1_2)

print("Переводы сохранены")
```

# **Оценка качества переводов моделей OPUS-MT и M2M-100 с помощью метрик BLEU и BERTScore**

```python
# 1. Устанавливаем библиотеки
!pip install sacrebleu
!pip install bert-score
!pip install pandas

# Импортируем инструменты для подсчета метрик
from sacrebleu.metrics import BLEU
from bert_score import score
import pandas as pd

# Загрузка эталона (ДЛЯ КАЖДОГО ЖАНРА БУДЕТ СВОЙ ЭТАЛОННЫЙ ПЕРЕВОД!!!!!)
with open("reference.txt", "r", encoding="utf-8") as f: # НЕ ЗАБУДЬТЕ ЗДЕСЬ ЗАМЕНИТЬ НАЗВАНИЕ ФАЙЛА ЭТАЛОННОГО ПЕРЕВОДА!!!!!!!!!!!!!!!!
    reference = f.read()

# Список файлов с переводами
files = {
    "OPUS-MT (Greedy)": "opus_greedy.txt",
    "OPUS-MT (Beam 5)": "opus_beam.txt",
    "OPUS-MT (T=0.7)": "opus_sample.txt",
    "OPUS-MT (T=1.2)": "opus_sample_1.2.txt",
    "M2M-100 (Greedy)": "m2m_greedy.txt",
    "M2M-100 (Beam 5)": "m2m_beam.txt",
    "M2M-100 (T=0.7)": "m2m_sample.txt",
    "M2M-100 (T=1.2)": "m2m_sample_1.2.txt",
}

translations = []
for name, filename in files.items():
    with open(filename, "r", encoding="utf-8") as f:
        translations.append((name, f.read()))

# BLEU
bleu = BLEU()
bleu_scores = [bleu.sentence_score(t, [reference]).score for _, t in translations]

# BERTScore
candidates = [t for _, t in translations]
references = [reference] * len(translations)
P, R, F1 = score(candidates, references, lang="en", batch_size=8)

# Таблица
df = pd.DataFrame({
    "Система": [name for name, _ in translations],
    "BLEU": bleu_scores,
    "BERTScore F1": [f1.item() for f1 in F1]
})
print(df)
df.to_csv("metrics_results.csv", index=False, encoding="utf-8-sig")
```
